#### Ejercicio 01. Análisis de datos básico con estructuras nativas de python.
Crea una lista con las calificaciones de 5 estudiantes: [8.5, 9.2, 7.8, 8.9, 10].

- Calcula el promedio de las calificaciones.

```python
# Código
estudiantes_calf=[8.5, 9.2, 7.8, 8.9, 10]
suma=0
for item in estudiantes_calf:
    suma+=item
    # comment: 
# end for
promedio=suma/(len(estudiantes_calf))
promedio
```

- Encuentra la calificación más alta y la más baja.

```python
# Código
cal_alta=0
cal_baja=estudiantes_calf[0]
for item in estudiantes_calf:
    if(item>cal_alta):
        cal_alta=item
    elif(item<cal_baja):
        cal_baja=item
    # comment: 
# end for
(cal_alta,cal_baja)
```
